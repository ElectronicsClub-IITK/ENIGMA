```python
import PySimpleGUI as sg
import serial
import time
```

```python
ser = serial.Serial('COM4', 115200, timeout=0.001)
```

### Task (for the code below):

- Add three more text input sections for rotor1, rotor2, and rotor3.

- Add a listbox to choose letters for the rotors from A

```python
font_spec = 'Courier 24 bold'
heading_color = '#2FB8AD'
alphabet_list = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")

# This defines how the GUI looks
layout = [
    [sg.Text('Please Input Characters', background_color=heading_color)],
    [sg.InputText('', key='pic_input'), sg.Button('Send', key='pic_send')],
    [sg.Text('Rotor 1', background_color=heading_color),sg.Text(' '*54),
    sg.Listbox(values=alphabet_list, size=(10, 1), enable_events=True, key='offset_1')],
    [sg.InputText(key="rotor_1"),sg.Button('Send', key='rotor1_send')],
    [sg.Text('Rotor 2',  background_color=heading_color),sg.Text(' '*54),
    sg.Listbox(values=alphabet_list, size=(10, 1), enable_events=True, key='offset_2')],
    [sg.InputText(key='rotor_2'),sg.Button('Send', key='rotor2_send')],
    [sg.Text('Rotor 3', background_color=heading_color),sg.Text(' '*54),
    sg.Listbox(values=alphabet_list, size=(10, 1), enable_events=True, key='offset_3')],
    [sg.InputText(key='rotor_3'),sg.Button('Send', key='rotor3_send')],
    [sg.Checkbox('Enable Reset', key='r_en'),sg.Button('Reset', key='rtg')],
]
```

### Apply Custom Theme to GUI
Task: Try changing colors! Make your GUI colorful! 

```python
sg.SetOptions(
    background_color='#9FB8AD',
    text_element_background_color='#9FB8AD',
    element_background_color='#475841',
    button_color=('white', '#475841'),
)
```

```python
##Create and show the window
window = sg.Window('ECE4760 Interface', layout, finalize=True)
while True:
    event, values = window.read(timeout=20)
    
    if event == sg.WIN_CLOSED or event == 'Exit':
        break

    # Send typed message to PIC
    if event == 'pic_send':
        message = values['pic_input']
        print(message)
        #ser.write((message + '\r').encode())
    if event == 'rotor1_send':
        settings= values['rotor_1']
        print(settings)
    if event == 'rotor2_send':
        settings= values['rotor_2']
        print(settings)
    if event == 'rotor3_send':
        settings= values['rotor_3']
        print(settings)
    if event == 'rtg' and values['r_en']:
        print('reset done')
        ser.send_break()
    
```
